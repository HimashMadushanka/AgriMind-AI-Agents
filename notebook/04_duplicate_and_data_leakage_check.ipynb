### Import Libraries

```python
from pathlib import Path
import pandas as pd
```

### Load Dataset

```python
ROOT = Path.cwd()

if ROOT.name == "notebooks":
    ROOT = ROOT.parent

df = pd.read_csv(
    ROOT / "../data/processed/cleaned_paddy_qa.csv",
    keep_default_na=False,
    encoding="utf-8-sig",
)
```

### Duplicate and leakage checks

```python
print(
    "Duplicate pair/language keys:",
    df.duplicated(["pair_id", "language"]).sum(),
)

print(
    "Duplicate question-answer-language rows:",
    df.duplicated(["question", "answer", "language"]).sum(),
)

print("\nRepeated questions by language:")
print(
    df.groupby("language")["question"]
    .apply(lambda values: values.duplicated().sum())
)

print("\nCategories per issue ID:")
print(
    df.groupby("issue_id")["category_en"].nunique()
    .value_counts()
)

assert (
    df.groupby("issue_id")["category_en"]
    .nunique()
    .max() == 1
)

print(
    "\nUse issue_id for splitting so variants of the same "
    "underlying issue stay in one split."
)
```

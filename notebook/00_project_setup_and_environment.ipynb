```python
from pathlib import Path
import sys
import importlib.util
import platform

# Project location
PROJECT_PATH = Path(
    r"D:\My_Project\Analyse\AI-Paddy-Farming-Intelligence"
)

# Locate project root
NOTEBOOK_DIR = (
    Path(__file__).resolve().parent
    if "__file__" in globals()
    else Path.cwd()
)

SEARCH_DIRS = [
    Path.cwd(),
    *Path.cwd().parents,
    NOTEBOOK_DIR,
    *NOTEBOOK_DIR.parents,
]

ROOT = next(
    (
        directory
        for directory in SEARCH_DIRS
        if directory == PROJECT_PATH
        and (
            (directory / "data").is_dir()
            or (directory / "datasets").is_dir()
        )
    ),
    PROJECT_PATH,
)

# Detect data directory
if (ROOT / "data").is_dir():
    DATA_DIR = ROOT / "data"
elif (ROOT / "datasets").is_dir():
    DATA_DIR = ROOT / "datasets"
else:
    raise FileNotFoundError(
        f"No data or datasets folder found in {ROOT}"
    )

# Display environment information
print("=" * 55)
print("AI Paddy Farming Intelligence - Environment Check")
print("=" * 55)

print("Project:", ROOT.resolve())
print("Data directory:", DATA_DIR.resolve())
print("Python executable:", sys.executable)
print("Python version:", sys.version.split()[0])
print("Platform:", platform.platform())

# Check installed libraries
packages = {
    "pandas": "pandas",
    "numpy": "numpy",
    "scikit-learn": "sklearn",
    "PyTorch": "torch",
    "Transformers": "transformers",
    "Sentence Transformers": "sentence_transformers",
    "FastAPI": "fastapi",
    "Streamlit": "streamlit",
    "Google GenAI": "google.genai",
}

print("\nPackage Status")
print("-" * 55)

for name, module in packages.items():
    try:
        installed = importlib.util.find_spec(module) is not None
    except (ImportError, ModuleNotFoundError, ValueError):
        installed = False

    print(f"{name}: {'Installed' if installed else 'Missing'}")

# Check PyTorch and GPU
print("\nPyTorch and GPU Status")
print("-" * 55)

try:
    import torch

    print("PyTorch version:", torch.__version__)
    print("CUDA available:", torch.cuda.is_available())

    if torch.cuda.is_available():
        print("GPU:", torch.cuda.get_device_name(0))
        print("CUDA version:", torch.version.cuda)
    else:
        print("GPU acceleration is unavailable. Using CPU.")

except ImportError:
    print("PyTorch is not installed in the active Python environment.")

print("\nEnvironment check completed.")
```

```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd()

if ROOT.name == "notebooks":
    ROOT = ROOT.parent

source = (
    ROOT / "../data/processed/multilingual_paddy_qa.csv"
)

df = pd.read_csv(
    source,
    encoding="utf-8-sig",
    keep_default_na=False,
)

required_columns = [
    "question",
    "answer",
    "language",
    "category_en",
    "issue_id",
    "pair_id",
]

for col in required_columns:
    assert col in df.columns, f"Missing column: {col}"

print("Original shape:", df.shape)
print("Empty values:")
print(df.isna().sum())

# Normalize whitespace.
for col in df.select_dtypes(include="object"):
    df[col] = df[col].astype(str).str.strip()

# Remove duplicate record keys.
df = df.drop_duplicates(
    subset=["pair_id", "language"],
    keep="first",
).copy()

# Remove invalid text rows.
df = df[
    df["question"].str.len().gt(0)
    & df["answer"].str.len().gt(0)
    & df["category_en"].str.len().gt(0)
].copy()

assert set(df["language"].unique()) == {
    "si-LK",
    "en-LK",
}

assert (
    df.groupby("issue_id")["category_en"]
    .nunique()
    .max() == 1
), "An issue maps to more than one category."

output = ROOT / "../data/processed/cleaned_paddy_qa.csv"

df.to_csv(
    output,
    index=False,
    encoding="utf-8-sig",
)

print("Cleaned shape:", df.shape)
print("Saved:", output)
```

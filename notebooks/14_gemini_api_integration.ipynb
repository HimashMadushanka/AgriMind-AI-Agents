# 14 — Gemini API Integration

## Goal
Make a minimal Gemini API call using Google's current `google-genai` Python SDK.

**What happens using this code?** It reads your API key from `.env` and sends a short test prompt. Do not put your real key in a notebook cell or commit it to Git.

```python
from pathlib import Path
import os
from dotenv import load_dotenv
ROOT = Path.cwd()
if ROOT.name == "notebooks": ROOT = ROOT.parent
load_dotenv(ROOT / ".env")
api_key = os.getenv("GEMINI_API_KEY")
if not api_key:
    print("Add GEMINI_API_KEY=your_key to the project .env file, then restart this cell.")
else:
    from google import genai
    client = genai.Client(api_key=api_key)
    model_name = os.getenv("GEMINI_MODEL", "gemini-3.8-flash")
    response = client.models.generate_content(model=model_name, contents="Explain the purpose of an agricultural NLP assistant in two simple sentences.")
    print(response.text)
```

**Current SDK:** the project uses `google-genai`, which is Google's recommended Python SDK for the Gemini API. Check Google's current model availability and usage limits before deployment: https://ai.google.dev/gemini-api/docs/libraries.

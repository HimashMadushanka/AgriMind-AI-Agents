# 12 — Search Expert-Reviewed Agricultural Documents

## Goal
Create a semantic-search index over documents placed in `knowledge_base/documents/`.

**What happens using this code?** The helper reads `.txt`, `.md`, and `.pdf` files, breaks them into overlapping chunks, and returns chunks ranked by multilingual semantic similarity. Only add trusted, legally usable, expert-reviewed material here.

```python
from pathlib import Path
import sys
ROOT = Path.cwd()
if ROOT.name == "notebooks": ROOT = ROOT.parent
if str(ROOT) not in sys.path: sys.path.insert(0, str(ROOT))
from tools.search_agricultural_knowledge import search_agricultural_knowledge

doc_dir = ROOT / "knowledge_base/documents"
files = [p.name for p in doc_dir.rglob("*") if p.is_file() and p.suffix.lower() in {".txt", ".md", ".pdf"}]
print("Supported documents currently found:", files)
if not files:
    print("Add official or expert-reviewed documents before testing retrieval.")
else:
    query = "වී වගාවේ රෝග ලක්ෂණ හඳුනාගැනීම"
    results = search_agricultural_knowledge(query, k=5)
    for item in results:
        print(f"\nScore: {item['score']:.3f} | Source: {item['source']}\n{item['text'][:500]}")
```

**What makes RAG useful?** Retrieval provides relevant source excerpts to the language model. It reduces unsupported answers only when the documents are authoritative, retrieval is checked, and the generated answer stays grounded in those excerpts.

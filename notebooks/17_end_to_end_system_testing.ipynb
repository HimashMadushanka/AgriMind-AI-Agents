## Goal
Run basic validation checks before deploying the prototype.

**What happens using this code?** It checks data integrity and the rule-based risk tool. These are smoke tests; you still need testing by farmers and agricultural officers.

```python
from pathlib import Path
import pandas as pd
import sys
ROOT = Path.cwd()
if ROOT.name == "notebooks": ROOT = ROOT.parent
if str(ROOT) not in sys.path: sys.path.insert(0, str(ROOT))
from tools.assess_farming_risk import assess_farming_risk

df = pd.read_csv(ROOT / "datasets/processed/multilingual_paddy_qa.csv", keep_default_na=False, encoding="utf-8-sig")
assert len(df) == 25440
assert df["pair_id"].nunique() == 12720
assert not df.duplicated(["pair_id", "language"]).any()
assert set(df["language"]) == {"si-LK", "en-LK"}
assert df.groupby("issue_id")["category_en"].nunique().max() == 1
print("Dataset integrity checks passed.")

case = assess_farming_risk("My field is severely flooded")
assert case["risk_level"] == "high"
print("Risk-rule smoke test passed:", case)
print("\nNOTE: the rules are examples only and must be reviewed by agricultural experts.")
```

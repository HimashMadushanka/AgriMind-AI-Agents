# 09 — Compare Models and Select a Candidate

## Goal
Compare the TF-IDF baseline and XLM-RoBERTa using validation metrics only.

**What happens using this code?** It selects the candidate with the better macro-F1 score (then accuracy as a tie-breaker) and saves that choice. Macro-F1 treats each category equally, which matters if category difficulty or distribution differs.

```python
from pathlib import Path
import json, pandas as pd
ROOT = Path.cwd()
if ROOT.name == "notebooks": ROOT = ROOT.parent
result_dir = ROOT / "models/model_evaluation_results"
files = [result_dir / "baseline_validation_metrics.json", result_dir / "xlmr_validation_metrics.json"]
rows = []
for path in files:
    if path.exists(): rows.append(json.loads(path.read_text(encoding="utf-8")))
    else: print("Not found yet:", path.name, "— run its training notebook first.")
if not rows: raise FileNotFoundError("Run notebook 07 and/or 08 first.")
comparison = pd.DataFrame(rows).sort_values(["f1_macro", "accuracy"], ascending=False)
display(comparison)
selected = comparison.iloc[0].to_dict()
selection = {"selected_model":selected["model"], "selection_metric":"validation macro-F1; accuracy breaks ties", "validation_metrics":selected}
(ROOT / "models/model_evaluation_results/selected_model.json").write_text(json.dumps(selection, indent=2), encoding="utf-8")
print("Selected candidate:", selection["selected_model"])
print("This is a validation decision; do not change it after looking at the final test score.")
```

If only one model has been trained, this notebook selects the available candidate. For a fair study, train both before making your final choice.

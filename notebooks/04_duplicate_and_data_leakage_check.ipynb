### Import libraries and Load dataset

```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd()

if ROOT.name == "notebooks":
    ROOT = ROOT.parent

df = pd.read_csv(
    ROOT / "datasets/processed/cleaned_paddy.csv",
    keep_default_na=False,
    encoding="utf-8-sig"
)
```

### Check duplicate pair-language keys

```python
print(
    "Duplicate pair_id+language keys:",
    int(df.duplicated(["pair_id", "language"]).sum())
)
```

### Check duplicate question-answer pairs

```python
print(
    "Duplicate exact question-answer pairs:",
    int(df.duplicated(["question", "answer", "language"]).sum())
)
```

```python
print("Repeated questions by language:")

print(
    df.groupby("language")["question"]
    .apply(lambda x: int(x.duplicated().sum()))
)
```

### Check issue-category consistency

```python
print(
    "Issue IDs mapping to multiple categories:",
    int(
        (
            df.groupby("issue_id")["category_en"].nunique() > 1
        ).sum()
    )
)
```

### Display repeated questions

```python
for language, sub in df.groupby("language"):

    repeated = sub[
        sub.duplicated("question", keep=False)
    ][
        ["language", "question", "answer", "issue_id", "category_en"]
    ]

    print(f"\nRepeated question examples: {language}")

    display(repeated.head(10))
```
